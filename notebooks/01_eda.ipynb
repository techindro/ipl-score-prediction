# Notebook 1 — Exploratory Data Analysis

This notebook explores the IPL ball-by-ball dataset to understand its structure,
distributions, and key patterns before modelling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
%matplotlib inline
```

## 1. Load the Dataset

```python
df = pd.read_csv('../data/ipl_dataset.csv')
print('Shape:', df.shape)
df.head()
```

```python
# Basic info
df.info()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Summary statistics
df.describe()
```

## 2. Matches per Venue

How many unique matches were played at each ground?

```python
matches_per_venue = df[['mid', 'venue']].drop_duplicates()
matches_count = matches_per_venue['venue'].value_counts()

plt.figure(figsize=(12, 6))
sns.barplot(x=matches_count.values, y=matches_count.index, palette='rainbow')
plt.title('Number of Matches per Venue', fontsize=14, fontweight='bold')
plt.xlabel('Number of Matches')
plt.ylabel('Venue')
plt.tight_layout()
plt.savefig('../outputs/matches_per_venue.png', dpi=150)
plt.show()
```

## 3. Top 10 Batsmen by Maximum Runs

```python
runs_by_batsman = (
    df.groupby('batsman')['runs']
    .max()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))
sns.barplot(x=runs_by_batsman.values, y=runs_by_batsman.index, palette='pastel')
plt.title('Top 10 Batsmen by Maximum Runs in an Innings', fontsize=14, fontweight='bold')
plt.xlabel('Maximum Runs')
plt.ylabel('Batsman')
plt.tight_layout()
plt.savefig('../outputs/top_batsmen.png', dpi=150)
plt.show()
```

## 4. Top 10 Bowlers by Maximum Wickets

```python
wickets_by_bowler = (
    df.groupby('bowler')['wickets']
    .max()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))
sns.barplot(x=wickets_by_bowler.values, y=wickets_by_bowler.index, palette='muted')
plt.title('Top 10 Bowlers by Maximum Wickets in an Innings', fontsize=14, fontweight='bold')
plt.xlabel('Maximum Wickets')
plt.ylabel('Bowler')
plt.tight_layout()
plt.savefig('../outputs/top_bowlers.png', dpi=150)
plt.show()
```

## 5. Distribution of Final Innings Totals

What does the target variable (total) look like?

```python
totals = df[['mid', 'total']].drop_duplicates()

plt.figure(figsize=(10, 5))
sns.histplot(totals['total'], bins=40, kde=True, color='steelblue')
plt.title('Distribution of Final Innings Totals', fontsize=14, fontweight='bold')
plt.xlabel('Total Runs')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('../outputs/total_distribution.png', dpi=150)
plt.show()

print(f"Mean total  : {totals['total'].mean():.1f}")
print(f"Median total: {totals['total'].median():.1f}")
print(f"Min total   : {totals['total'].min()}")
print(f"Max total   : {totals['total'].max()}")
```

## 6. Team-wise Average Total

Which teams tend to score higher?

```python
team_totals = (
    df[['mid', 'bat_team', 'total']]
    .drop_duplicates()
    .groupby('bat_team')['total']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 5))
sns.barplot(x=team_totals.index, y=team_totals.values, palette='Set2')
plt.title('Average Innings Total by Team', fontsize=14, fontweight='bold')
plt.xlabel('Team')
plt.ylabel('Average Total')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('../outputs/team_avg_total.png', dpi=150)
plt.show()
```
